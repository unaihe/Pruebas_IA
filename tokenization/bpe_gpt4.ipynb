```python
%pip install regex
import regex 

class BasicTokenizer:
    def __init__(self):
        self.patron = r"""'(?i:[sdmt]|ll|ve|re)|[^\r\n\p{L}\p{N}]?+\p{L}+|\p{N}{1,3}| ?[^\s\p{L}\p{N}]++[\r\n]*|\s*[\r\n]|\s+(?!\S)|\s+"""
        self.merges = {}
     
    # Importamos la función merge de la fase anterior del video
    def merge(self, ids, par, nuevo_id):
        i = 0
        l_ids = len(ids)
        nuevo_ids = []
        
        while i <= l_ids - 1:
            if i == l_ids - 1:
                nuevo_ids.append(ids[i])
                i += 1
            elif (ids[i], ids[i + 1]) == par:
                nuevo_ids.append(nuevo_id)
                i += 2
            else:
                nuevo_ids.append(ids[i])
                i += 1

        # Se devuelve nuevos_ids que es ids modificado de tal forma que en vez de la pareja de tokens aparezca el nuevo_id
        return nuevo_ids
    
    def train(self, text, vocab_size, verbose=False):
        # Se trata de la función de donde se sacarán los merges que usaremos en el futuro
        # Primero de todo hay que separar el texto con el regex
        sep = regex.findall(self.patron, text)
        # Tras separarlo habría que pasar todos los valores a nivel de byte
        encoded_text = []
        for fragmento in sep: 
            encoded_text.append(list(fragmento.encode("UTF-8")))
        
        # Tras eso debemos de sacar los pares mas comunes, y crear el vacbulario con su diccionario de merges
        # Metemos el conteo en el interior debido a que cada vez que se hace una fusión nueva se debe de recalcular
        i = 256
        
        while i < vocab_size:
            # Una vez tenemos ya todo codificado en utf-8 buscamos los pares más ocurrentes
            # Al ser encoded_text una lista de listas debemos de crear dos bucles para poder iterar por cada byte.
            countPar = {}
            for word in encoded_text:
                for par in zip(word, word[1:]):
                    countPar[par] = countPar.get(par, 0) + 1
            if len(countPar) == 0:
                break
            parMaximo = max(countPar, key = countPar.get)
            self.merges[parMaximo] = i
           

            nuevo_encoded = []
            for word in encoded_text:
                nuevo_encoded.append(self.merge(word, parMaximo, i))
            encoded_text = nuevo_encoded
            i += 1
            
    def encode(self, text):
        pass

    def decode(self, ids):
        pass

bpe = BasicTokenizer()
text = "Hpoñla que tal?? , yo muy bien."
patron = r"""'(?i:[sdmt]|ll|ve|re)|[^\r\n\p{L}\p{N}]?+\p{L}+|\p{N}{1,3}| ?[^\s\p{L}\p{N}]++[\r\n]*|\s*[\r\n]|\s+(?!\S)|\s+"""
sep = regex.findall(patron, text)
print(sep)
encoded_text = []
for fragmento in sep: 
    encoded_text.append(list(fragmento.encode("UTF-8")))
print(encoded_text)
```
