```python
# Using the same text as in the video
text = "Ｕｎｉｃｏｄｅ! 🅤🅝🅘🅒🅞🅓🅔‽ 🇺‌🇳‌🇮‌🇨‌🇴‌🇩‌🇪! 😄 The very name strikes fear and awe into the hearts of programmers worldwide. We all know we ought to “support Unicode” in our software (whatever that means—like using wchar_t for all the strings, right?). But Unicode can be abstruse, and diving into the thousand-page Unicode Standard plus its dozens of supplementary annexes, reports, and notes can be more than a little intimidating. I don’t blame programmers for still finding the whole thing mysterious, even 30 years after Unicode’s inception."
tokens = text.encode('utf-8')
# Se usa list para mostrar los elementos dentro del map (realmente el map no es necesario porque el encode devuelve números enteros) 
# Es decir con tokens = list(tokens) sale el mismo resultado
tokens = list(map(int, tokens))
print(tokens)
print(len(tokens))
print("--------------")

# Ahora debemos de calcular cuales son los bytes que más aparecen de seguido
# Planteamiento crear un diccionario donde se vayan guardando los dos valores y las veces que aparecen
# Se trata de una implementación propia (seguro que hay mas efectivas)
def get_stats(tokens):
    dic = {}
    for idx in range(len(tokens) - 1):
        par = (tokens[idx], tokens[idx + 1])
        
        if par in dic:
            dic[par] += 1
        else:
            dic[par] = 1

    return dic

"""
Manera en la que lo hace en el video:

def get_stats(ids):
    counts = {}
    for pair in zip(ids, ids[1:]):
        counts[pair] = counts.get(pair, 0) + 1
    return counts

"""

dic = get_stats(tokens)
maxTuple = max(dic, key=dic.get)
print(maxTuple, dic[maxTuple])
print("--------")

def merge(ids, par, nuevo_id):
    i = 0
    l_ids = len(ids)
    nuevo_ids = []
    
    while i <= l_ids - 1:
        if i == l_ids - 1:
            nuevo_ids.append(ids[i])
            i += 1
        elif (ids[i], ids[i + 1]) == par:
            nuevo_ids.append(nuevo_id)
            i += 2
        else:
            nuevo_ids.append(ids[i])
            i += 1

    # Se devuelve nuevos_ids que es ids modificado de tal forma que en vez de la pareja de tokens aparezca el nuevo_id
    return nuevo_ids
    
nuevo_tokens = merge(tokens, maxTuple, 256)
print(nuevo_tokens)
print(len(nuevo_tokens))

# Ahora tratare de implementar el bucle para que se tokenicen los 20 pares mas frecuentes y el vocabulario por lo tanto deberia de ser 275
def tokenization(tokens, fusiones):
    maxTuplesId = {}
    
    for i in range(fusiones):
        dic = get_stats(tokens)
        maxTuple = max(dic, key=dic.get)
        nuevo_id = 256 + i
        tokens = merge(tokens, maxTuple, nuevo_id)
        maxTuplesId[maxTuple] = nuevo_id

    return maxTuplesId, tokens



```
